```python
#Operators are used to perform mathematical operations.

print("--- Operators ---")

a = 10
b = 5

print("Addition:", a + b) # Adds two numbers
print("Subtraction:", a - b) # Subtracts
print("Multiplication:", a * b) # Multiplies
print("Division:", a / b) # Divides (returns float)
print("Modulus:", a % b) # Remainder
```

### even_odd.py


```python
# Check whether a number is even or odd

num = int(input("Enter a number: "))
if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

### largest_number.py


```python
# Find the largest of three numbers

a = int(input("Enter first number: ")) 
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))
if a >= b and a >= c:
    print("Largest:", a)
elif b >= a and b >= c:
    print("Largest:", b)
else:
    print("Largest:", c)
```

### factorial.py

```python
# Find factorial of a number

num = int(input("Enter a number: "))
fact = 1
for i in range(1, num + 1):
    fact *= i
print("Factorial:", fact)
```

### fibonacci.py

```python
# Print Fibonacci series

n = int(input("Enter number of terms: "))
a, b = 0, 1
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

### palindrome.py

```python
# Check if a number is palindrome

num = input("Enter a number: ")
if num == num[::-1]:  
    print("Palindrome")
else:
    print("Not Palindrome")
```

### sum_of_list.py

```python
# Find sum of elements in a list

numbers = [1, 2, 3, 4, 5]
total = sum(numbers)
print("Sum of list:", total)
```

### prime_number.py

```python
# Check whether a number is prime

num = int(input("Enter a number: "))
if num > 1:
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
else:
    print("Not Prime")
```
